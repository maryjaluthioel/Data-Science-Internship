```python
#1d
import numpy as np
arr1d = np.array([1, 2, 3, 4])
print(arr1d)
print(arr1d.ndim)
#access
print(arr1d[2])
```

```python
#2d
import numpy as np
arr2d = np.array([[1, 2, 3], [4, 5, 6]])
print(arr2d)
print(arr2d.ndim)
#access
print(arr2d[0,2])
```

```python
#3d
import numpy as np
arr3d = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print(arr3d)
print(arr3d.ndim)
print(arr3d)
```

```python
#MATPLOTLIB LINE PLOT
import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]
plt.plot(x, y, color='blue', marker='o')
plt.title("Basic Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
#SCATTER PLOT
import pandas as pd
from pandas.plotting import scatter_matrix
import matplotlib.pyplot as plt
df = pd.DataFrame({
"Height": [150, 160, 170, 180, 190],
"Weight": [50, 60, 70, 80, 90],
"Age": [20, 25, 30, 35, 40]
})
scatter_matrix(df, alpha=0.8, figsize=(6, 6), diagonal="hist")
plt.suptitle("Scatter Matrix Example", size=14)
plt.show()
```

```python
#LAG POT
import pandas as pd
import matplotlib.pyplot as plt
from pandas.plotting import lag_plot
ts = pd.Series([10, 12, 15, 20, 22, 25, 30, 35, 40, 42, 45])
lag_plot(ts)
plt.title("Lag Plot Example")
plt.show()
```

```python
#AUTOCORR
import pandas as pd
import matplotlib.pyplot as plt
from pandas.plotting import autocorrelation_plot
ts = pd.Series([10, 12, 15, 20, 22, 25, 30, 35, 40, 42, 45])
autocorrelation_plot(ts)
plt.title("Autocorrelation Plot Example")
plt.show()
```

```python
#BOOTSTRAP
import pandas as pd
import matplotlib.pyplot as plt
from pandas.plotting import autocorrelation_plot
ts = pd.Series([10, 12, 15, 20, 22, 25, 30, 35, 40, 42, 45])
autocorrelation_plot(ts)
plt.title("Autocorrelation Plot Example")
plt.show()
```

```python
import matplotlib.pyplot as plt
subjects = ["Python", "Java", "C++", "PHP"]
marks = [85, 78, 90, 70]
plt.bar(subjects, marks, color = "gray")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Student Marks Bar Plot")
plt.show()
```

```python
#pie chart
import matplotlib.pyplot as plt
subjects = ["Python", "Java", "C++", "PHP"]
marks = [85, 78, 90, 70]
plt.pie(marks, labels=subjects, autopct="%1.1f%%", colors=["red", "darkgray", "darkred", "silver"])
plt.title("Student Marks Pie Chart")
plt.show()
```
