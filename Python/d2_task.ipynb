```python
#5.
class LoginError(Exception):
  pass
crct_psw = "1234"
attempts = 0
while attempts <  3:
  try:
    psw = input("enter psw:")
    if psw != crct_psw:
       attempts += 1
       raise LoginError("wrong psw")
    print("login successfull")
    break
  except LoginError as e:
    print(e)
else:
  print("account locked after 3 attempts")
```

```python
#8.
class InvalidAge(Exception):
  pass
try:
    age = int(input("enter the age:"))
    if age < 18:
       raise InvalidAge("age must be 18 or above")
    print("successful")
except InvalidAge as e:
    print(e)
except ValueError:
    print("value error")
```

```python
#9.
try:
  try:
    a = int(input("enter a:"))
    b = input("enter b:")
  except ValueError:
    print("val error")
  else:
    try:
      if b.isdigit():
        b = int(b)
      result = a / b
      print("res:",result)
    except ZeroDivisionError:
      print("zero div error")
    except TypeError:
      print('type error')
finally:
  print("program executed")
```

```python
#7.
name_list = ["jaa","her","eps","rus","swa","har","nan","dd","jan","anu","her","jaa"]
with open("names.txt", "w") as file:
    for name in name_list:
        file.write(name + "\n")
with open("names.txt", "r") as file:
    lines = file.readlines()
unique_lines = list(dict.fromkeys(lines))
with open("names.txt", "w") as file:
    file.writelines(unique_lines)
print("--- Content After Removing Duplicates ---")
for line in unique_lines:
    print(line.strip())
```

```python
#8.
student_data = "Anu,85\n Rushiba,92\n swany,78"
with open("marks.txt", "w") as file:
    file.write(student_data)
total = 0
count = 0
with open("marks.txt", "r") as file:
    for line in file:
        name, mark = line.strip().split(",")
        total += int(mark)
        count += 1
print(f"Average Mark: {total / count}")
```

```python
#9.
data_list = []
with open("marks.txt", "r") as file:
    for line in file:
        row = line.strip().split(",")
        data_list.append({"name": row[0], "score": int(row[1])})

print("Parsed Data:", data_list)
```

```python
#10.
updated_lines = []
with open("marks.txt", "r") as file:
    for line in file:
        if line.startswith("Rushiba"):
            updated_lines.append("Rushiba,95\n")
        else:
            updated_lines.append(line)
with open("marks.txt", "w") as file:
    file.writelines(updated_lines)
print("--- Updated marks.txt Content ---")
with open("marks.txt", "r") as file:
    print(file.read())
```

```python
#10.
def divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Cannot divide by zero"
print(divide(10, 2))
print(divide(10, 0))
```

```python
#10.
with open("errors.py", "w") as f:
    f.write("""def divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Cannot divide by zero"
""")
import errors
print("From errors module:")
print(errors.divide(10, 2))
print(errors.divide(10, 0))
```
