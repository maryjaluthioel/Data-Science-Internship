```python
import random
words = ["python", "table", "chair","check","robot"]
word = random.choice(words)
display = ["_"] * len(word)
guessed = []
wrong = 0
stages = [
r"""-----
|   |
    |
    |
    |
    |
=========""",
r"""-----
|   |
O   |
    |
    |
    |
=========""",
r"""-----
|   |
O   |
|   |
    |
    |
=========""",
r"""-----
|   |
O   |
/|  |
    |
    |
=========""",
r"""-----
|   |
O   |
/|\ |  # Corrected: Was /|\ | but raw string should handle. Warning might be from '`\ `'
    |
    |
=========""",
r"""-----
|   |
O   |
/|\ |  # Corrected: Was /|\ |
/   |
    |
=========""",
r"""-----
|   |
O   |
/|\ |  # Corrected: Was /|\ |
/ \ |  # Corrected: Changed / \ | to / \\ |
    |
========="""
]
while wrong < 6 and "_" in display:
    print(stages[wrong])
    print("\nWord:", " ".join(display))
    guess = input("\nEnter a letter: ")
    if guess in guessed:
        continue
    guessed.append(guess)
    if guess in word:
        for i in range(len(word)):
            if word[i] == guess:
                display[i] = guess
    else:
        wrong += 1
        print("\nWrong guess! Attempts left:", 6 - wrong)
if "_" not in display:
    print("\nWord:", " ".join(display))
else:
    print(stages[wrong])
    print("\nWord:", " ".join(display))
```
