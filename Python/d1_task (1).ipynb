```python
#4.count vowels conso
vowels = "aeiou"
vowel_count = sum(1 for char in a if char in vowels)
print("vowels:", vowel_count)
print("cons:", len(a) - vowel_count)
```

```python
#4.count vowels conso
vowels = "aeiou"
v=c=0
for ch in a:
  if ch in vowels:
    v+=1
  else:
    c+=1
print("vowels:",v)
print("cons:",c)
```

```python
#8.freq count(cant use count() in str)
s = "hello jalu"
freq = {}
for ch in s:
  if ch in freq:
    freq[ch] += 1
  else:
    freq[ch] = 1
print(freq)
```
